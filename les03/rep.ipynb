````python
import numpy as np
np.version.version
````

1. Создать массив 10x10 со случайными значениями, найти минимум и максимум

````python
arr = np.random.randint(0, 100, (10, 10))

print(arr)
print("Минимум:", arr.min())
print("Максимум:", arr.max())
````

2. Создать 8x8 матрицу и заполнить её единицами и нулями в шахматном порядке

````python
arr = np.zeros((8, 8), dtype=int)
arr[1::2, ::2] = 1
arr[::2, 1::2] = 1

print(arr)
````

3. Дан массив из 11 элементов, поменять знак у элементов, значения которых между 3 и 8

````python
z = np.arange(11)

for i in range(len(z)):
    if 3 < z[i] < 8:
        z[i] = -z[i]

print(z)
````

4. Найти среднее заначение строки матрицы и вычесть его из каждого елемента строки

````python
z = np.random.rand(5, 10)
print(z)

for i in range(z.shape[0]):
    avg = z[i].mean()
    z[i] = z[i] - avg

print(z)
````

5. Поменяйте 1 и 2 строки матрицы

````python
z = np.arange(25).reshape(5,5)
z[[0, 1]] = z[[1, 0]]

print(z)
````

6. Дан массив 10x2 подеставлющий собой точки в декартовой системе координат, преобразовать в полярную

````python
z = np.random.random((10,2))
x = z[:, 0]
y = z[:, 1]

r = np.sqrt(x**2 + y**2)
theta = np.arctan2(y, x)

polar = np.column_stack((r, theta))

print("Декартовы:\n", z)
print("Полярные (r, theta):\n", polar)
````

7. Заменить максимальный элемент массива на ноль

````python
z = np.random.random(10)
print(z)

max_index = np.argmax(z)

z[max_index] = 0

print(z)
````

8. Дан массив, добавить 1 к каждому элементу с индексом, заданным в другом массиве. Добавление производиться только 1 раз

````python
z = np.ones(10)
i = np.random.randint(0,len(z),20)
print("Индексы", i)
unique_indices = np.unique(i)

z[unique_indices] += 1

print("Итог:", z)
````

9. Дан вектор [1, 2, 3, 4, 5], построить новый вектор с тремя нулями между каждым значением

````python
z = np.array([1,2,3,4,5])
n_zeros = 3

new_length = len(z) + (len(z) - 1) * n_zeros
arr = np.zeros(new_length, dtype=int)

arr[:: n_zeros + 1] = z

print(arr)
````

10. Найти ближайшее к заданному значению число в заданном массиве

````python
z = np.arange(100) # Число
v = np.random.uniform(0,100) # Массив
index = np.abs(z - v).argmin()

closest = z[index]

print("Число:", v)
print("Ближайшее:", closest)
````

**10**.  В чем преимущество массивов NumPy по сравнению с обычными списками.  Чем обусловлено это преимущество?

Преимущество NumPy: быстрее, меньше памяти, удобные векторные операции.

Причина: данные хранятся в одном непрерывном блоке памяти одного типа, что позволяет NumPy выполнять операции на С/Fortran уровне, минуя циклы Python.
