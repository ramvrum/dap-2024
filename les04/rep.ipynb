6. Правда ли, что люди моложе 30 лет выживали чаще, чем люди старше 60 лет? Каковы доли выживших в обеих группах?

````python
young = df[df['Age'] < 30]
young_ratio = round(young['Survived'].mean(), 2)

old = df[df['Age'] > 60]
old_ratio = round(old['Survived'].mean(), 2)

print("Моложе 30: ", young_ratio)
print("Старше 60: ", old_ratio)
````

7. Правда ли, что женщины выживали чаще мужчин? Каковы доли выживших в обеих группах?

````python
female = df[df['Sex'] == 'female']
female_ratio = round(female['Survived'].mean(), 2)

male = df[df['Sex'] == 'male']
male_ratio = round(male['Survived'].mean(), 2)

print("Женщины: ", female_ratio)
print("Мужчины: ", male_ratio)
````

### Дополнительные вопросы для защиты

1. Что такое медиана? Рассчитать медиану для предложенного ряда чисел.

Медиана — это число, которое делит ряд данных на две равные части.
Пример для ряда [1, 3, 2, 5, 4]:


```
import numpy as np
data = [1, 3, 2, 5, 4]
np.median(data)  # 3
```

2. Что такое дисперсия? Рассчитать дисперсию для предложенного ряда чисел.

Дисперсия — это мера разброса данных относительно среднего.

Если дисперсия большая → числа сильно разбросаны.

Если дисперсия маленькая → числа близки к среднему.

```
import numpy as np
data = [1, 3, 2, 5, 4]
np.var(data) #2.0
```

3. Что такое среднеквадратичное отклонение? Рассчитать среднеквадратичное отклонение для предложенного ряда чисел.

Это квадратный корень из дисперсии, показывает среднее отклонение элементов от среднего.

```
np.std(data)  # 1.414...
```
